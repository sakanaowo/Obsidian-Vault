### 2. Hadamard product vs matrix duplication

```python
# Hadamard
A = torch.arange(4).reshape(2, 2)
print(A)
B = torch.arange(4, 8).reshape(2, 2)
print(B)
```

```python
A * B
```

```python
# Matrix multiplication
torch.mm(A, B)
```
