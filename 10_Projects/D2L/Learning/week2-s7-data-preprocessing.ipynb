```python
import pandas as pd


df = pd.read_csv("../house-prices-advanced-regression-techniques/train.csv")
df.head()
```

```python
print("\033[38;5;208mPercentages of missing values:\033[0m\n")

nulls = False

for col in df.columns:
    z =  df[col].isnull().sum()
    if z != 0:
        print(f"{col}: {z*100/1460:.2f}%")
        nulls = True
        
if not nulls:
    print("There aren't any null values in the train dataset.")
```
