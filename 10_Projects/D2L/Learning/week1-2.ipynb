## Element-wise operations

```python
x = torch.tensor([1, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])

print(f"{x} + {y} = {x + y}")

print(f"{x} - {y} = {x - y}")

print(f"{x} * {y} = {x * y}")

print(f"{x} / {y} = {x / y}")

print(f"{x} ** {y} = {x ** y}")
```

## Concatenation

## Broadcasting

```python
a = torch.arange(3).reshape((3, 1))
b = torch.arange(2).reshape((1, 2))
a, b
```

```python
a+b

# a:        b:
# [[0],     [[0, 1]]
#  [1],
#  [2]]

# Sau broadcasting:
# a → [[0, 0],    b → [[0, 1],
#      [1, 1],         [0, 1],
#      [2, 2]]         [0, 1]]

a + b
# tensor([[0, 1],
#          [1, 2],
#          [2, 3]])
```
