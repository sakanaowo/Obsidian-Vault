```python
def preprocess_raw(text):
    text = text.replace("\u202f", " ").replace("\xa0", " ").lower()
    text = text.lower()
    out = []
    for i, char in enumerate(text):
        if i > 0 and re.match(r"[!,?\.]+", char):
            out.append(" " + char)
        else:
            out.append(char)
    return "".join(out)
```
