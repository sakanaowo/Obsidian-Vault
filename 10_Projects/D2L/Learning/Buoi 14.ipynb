# Giải thích công thức Softmax

> [!NOTE] ELI5
> Softmax biến "điểm thô" của mô hình thành xác suất.
> Điểm cao hơn sẽ thành xác suất cao hơn.
> Sau khi biến đổi, các xác suất luôn dương và cộng lại đúng bằng 1.

## Công thức

$$
\hat{y}_i = \mathrm{softmax}(\mathbf{o})_i = \frac{e^{o_i}}{\sum_{j=1}^{q} e^{o_j}}
$$

## Ý nghĩa từng thành phần

- $\mathbf{o} = (o_1, o_2, ..., o_q)$: vector **logits** (điểm thô) cho $q$ class.
- $e^{o_i}$: làm mọi giá trị thành số dương và khuếch đại chênh lệch giữa các logits.
- $\sum_{j=1}^{q} e^{o_j}$: hằng số chuẩn hóa để tổng các đầu ra bằng 1.
- $\hat{y}_i$: xác suất mô hình gán cho class $i$.

A đúng vì B, được thể hiện qua C:
- **A:** Softmax tạo ra phân phối xác suất hợp lệ.
- **B:** Tử số luôn dương và mẫu số là tổng các số dương.
- **C:** $\hat{y}_i > 0$ và $\sum_i \hat{y}_i = 1$.

## Ví dụ nhanh

Cho logits $\mathbf{o} = (2.0, 1.0, 0.1)$:

- $e^{2.0} \approx 7.39$
- $e^{1.0} \approx 2.72$
- $e^{0.1} \approx 1.11$
- Tổng $\approx 11.22$

Suy ra:

- $\hat{y}_1 = 7.39 / 11.22 \approx 0.659$
- $\hat{y}_2 = 2.72 / 11.22 \approx 0.242$
- $\hat{y}_3 = 1.11 / 11.22 \approx 0.099$

=> Mô hình nghiêng về class 1 với xác suất cao nhất.
